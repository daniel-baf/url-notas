```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/netflix_titles.csv')

df
```

```python
df['date_added'] = df['date_added'].str.strip()
df['date_added'] = pd.to_datetime(df['date_added'])
df['date_added'] = df['date_added'].dt.strftime('%Y-%m-%d')
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['date_added_year'] = df['date_added'].dt.year.fillna(0).astype(int)
df['date_added_month'] = df['date_added'].dt.month.fillna(0).astype(int)
df['date_added_day'] = df['date_added'].dt.day.fillna(0).astype(int)
df
```
